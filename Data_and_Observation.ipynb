```python
class BankAccount:
    def __init__(self, account_holder, balance):
        self.__balance = balance 
        self.account_holder = account_holder

    def deposit(self, amount):
        """Allows controlled modification of the private balance."""
        self.__balance += amount
        return f"New Balance: {self.__balance}"

    def get_balance(self):
        """Allows safe retrieval of the private balance."""
        return f"Balance: {self.__balance}"

account = BankAccount("John Doe", 1000)

print(account.deposit(500))
print(account.get_balance())
```

```python
class Animal:
    def speak(self):
        """Base method in the parent class."""
        return "Some sound"

class Dog(Animal):
    def speak(self):
        """Overriding the base method in the subclass."""
        return "Bark!"

dog = Dog()

print(dog.speak())
```

```python
class Vehicle:
    def move(self):
        """Method in the grandparent class."""
        return "Vehicle is moving"

class Car(Vehicle):
    def fuel_type(self):
        """Method in the parent class."""
        return "Uses gasoline"

class ElectricCar(Car):
    def fuel_type(self):
        """Overridden method in the third-level subclass."""
        return "Uses electricity"

tesla = ElectricCar()

print(tesla.move())

print(tesla.fuel_type())
```

```python
class Box:
    def __init__(self, weight):
        self.weight = weight

    def __add__(self, other):
        return Box(self.weight + other.weight)

box1 = Box(5)
box2 = Box(10)
box3 = box1 + box2 
output_7 = f"Total weight: {box3.weight} kg"
print(output_7)
```

```python
from abc import ABC, abstractmethod

class Appliance(ABC):
    @abstractmethod
    def turn_on(self):
        pass

class Fan(Appliance):
    def turn_on(self):
        return "Fan is spinning"

fan = Fan()
print(fan.turn_on())
```

```python
class Engine:
    def start(self):
        return "Engine started"

class Wheels:
    def roll(self):
        return "Wheels rolling"

class Car(Engine, Wheels):
    def drive(self):
        return "Car is moving"

my_car = Car()
print(my_car.start())
print(my_car.roll())
print(my_car.drive())
```

```python
class MathOperations:
    @staticmethod
    def square(num):
        return num * num

try:
    number = int(input("Enter a number: "))
    print("Square:", MathOperations.square(number))
except ValueError:
    print("Invalid input! Please enter an integer.")
```

```python
class ATM:
    def __init__(self, balance):
        self.balance = balance

    def withdraw(self, amount):
        if amount > self.balance:
            return "Insufficient balance"
        
        self.balance -= amount
        return f"Withdrawn {amount}, Remaining balance: {self.balance}"

atm = ATM(1000)
print(atm.withdraw(200))
```

```python
class SmartDevice:
    def __init__(self, device_name, status="Off"):
        self.device_name = device_name
        self.status = status

    def turn_on(self):
        self.status = "On"
        return f"{self.device_name} is now {self.status}"

light = SmartDevice("Living Room Light")
print(light.turn_on())
```

```python
class Library:
    def __init__(self):
        self.books = []

    def add_book(self, title):
        self.books.append(title)
        return f"Book '{title}' added to the library"

    def show_books(self):
        if not self.books:
            return "The library is currently empty."
        return f"Available Books: {', '.join(self.books)}"

library = Library()
print(library.add_book("Python Programming"))
print(library.show_books())
```
